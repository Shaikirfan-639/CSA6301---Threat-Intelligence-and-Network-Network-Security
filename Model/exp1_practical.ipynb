```python
text = input("Enter business scenario: ").lower()

if "bank" in text or "fraud" in text:
    print("Banking Fraud Prevention")
elif "hospital" in text or "health" in text:
    print("Healthcare Data Protection")
elif "ecommerce" in text or "shopping" in text:
    print("E-commerce Fraud Prevention")
else:
    print("General Threat Intelligence")
```
