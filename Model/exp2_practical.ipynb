```python
import socket

host = input("Enter host: ")

for port in [21, 22, 23, 25, 53, 80, 443]:
    s = socket.socket()
    s.settimeout(0.5)

    result = s.connect_ex((host, port))

    if result == 0:
        print("Port", port, "is OPEN")
    else:
        print("Port", port, "is CLOSED")

    s.close()
```
