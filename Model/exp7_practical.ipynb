```python
logs = [
    "ERROR login failed",
    "INFO user login",
    "ERROR server down"
]

events = []

for log in logs:
    level, message = log.split(" ", 1)
    events.append({"level": level, "message": message})

print("Structured Events:")

for event in events:
    print(event)

count = 0

for event in events:
    if event["level"] == "ERROR":
        count += 1

print("ERROR count:", count)
```
