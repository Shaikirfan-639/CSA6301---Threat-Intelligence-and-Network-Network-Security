```python
from urllib.parse import urlparse

url = input("Enter URL: ").lower()
score = 0

if "login" in url or "verify" in url or "free" in url:
    score += 1

host = urlparse(url).hostname

if host and host.replace(".", "").isdigit():
    score += 1

if url.count(".") > 3:
    score += 1

if score >= 2:
    print("Possible Phishing URL")
else:
    print("URL looks normal")
```
